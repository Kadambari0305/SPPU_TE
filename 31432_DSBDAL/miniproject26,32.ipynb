```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

pd.read_csv(...) → Reads CSV file and stores it in dataframe df.
df.head() → Displays first 5 rows of dataset to check data.

```python
df = pd.read_csv("covid_vaccine_statewise.csv")
df.head()
```

df['State'] != 'India' → Selects all rows where state is NOT "India".
df = df[...] → Updates dataframe by removing "India" row.

```python
# Remove India row
df = df[df['State'] != 'India']
```

Data Cleaning
Usually dataset has date-wise entries, so we take the latest record of each state.

Converts "Updated On" column into date format.
Useful for time-based analysis.
Groups data by each state.
sum() → Adds all vaccination values for each state.

```python
df = df.sort_values("Updated On")

latest_data = df.groupby("State").last().reset_index()

latest_data.head()
```

## B):- Number of Persons State-Wise Vaccinated for First Dose 

```python
first_dose = latest_data[["State","First Dose Administered"]]

print(first_dose)
```

plt.figure(figsize=(12,6)) → Sets graph size.
sns.barplot(...) → Creates bar chart.
x → State names
y → Total doses
plt.xticks(rotation=90) → Rotates labels for readability.
plt.show() → Displays graph.

```python
plt.figure(figsize=(12,6))
sns.barplot(x="State", y="First Dose Administered", data=first_dose)

plt.xticks(rotation=90)
plt.title("State Wise First Dose Vaccination in India")
plt.show()
```

## C):- Number of Persons State-Wise Vaccinated for Second Dose 

```python
second_dose = latest_data[["State","Second Dose Administered"]]

print(second_dose)
```

```python
plt.figure(figsize=(12,6))
sns.barplot(x="State", y="Second Dose Administered", data=second_dose)

plt.xticks(rotation=90)
plt.title("State Wise Second Dose Vaccination in India")
plt.show()
```

## Male vs Female Vaccination Comparison 

```python
gender_total = [
    latest_data["Male (Doses Administered)"].sum(),
    latest_data["Female (Doses Administered)"].sum()
]

labels = ["Male","Female"]

plt.pie(gender_total, labels=labels, autopct='%1.1f%%')
plt.title("Gender Wise Vaccination")
plt.show()
```

## Top 10 States with Highest First Dose Vaccination

## Top 10 States with Highest Second Dose Vaccination

```python
top10_second = latest_data.sort_values("Second Dose Administered", ascending=False).head(10)

plt.figure(figsize=(10,6))
sns.barplot(x="Second Dose Administered", y="State", data=top10_second)

plt.title("Top 10 States with Highest Second Dose Vaccination")
plt.xlabel("Second Dose Administered")
plt.ylabel("State")
plt.show()
```

## Male vs Female Vaccination Comparison

```python
gender = latest_data[["Male (Doses Administered)", "Female (Doses Administered)"]].sum()

plt.figure(figsize=(6,5))
gender.plot(kind='bar')

plt.title("Male vs Female Vaccination in India")
plt.ylabel("Number of Vaccinations")
plt.xticks(rotation=0)
plt.show()
```

## Heatmap of Vaccination Data

```python
heatmap_data = latest_data[[
    "First Dose Administered",
    "Second Dose Administered",
    "Male (Doses Administered)",
    "Female (Doses Administered)"
]]

plt.figure(figsize=(8,6))
sns.heatmap(heatmap_data.corr(), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap of Vaccination Data")
plt.show()
```

## Total Vaccination by State

```python
latest_data["Total Vaccination"] = latest_data["First Dose Administered"] + latest_data["Second Dose Administered"]

top_states = latest_data.sort_values("Total Vaccination", ascending=False).head(10)

top_states.set_index("State")[["First Dose Administered","Second Dose Administered"]].plot(
    kind="bar", stacked=True, figsize=(10,6)
)

plt.title("Total Vaccination (First + Second Dose) in Top States")
plt.ylabel("Vaccination Count")
plt.xticks(rotation=45)
plt.show()
```

Vaccination Trend Over Time

This project uses Python libraries like Pandas, NumPy, Matplotlib, and Seaborn to analyze COVID vaccination data. It involves data cleaning, grouping, sorting, and visualization to understand vaccination trends across states and over time.

```python
import matplotlib.pyplot as plt
import seaborn as sns


gender_data = df.groupby("State")[["Male (Doses Administered)", "Female (Doses Administered)"]].sum().reset_index()

# Convert into long format
gender_data_melted = gender_data.melt(id_vars="State",
                                      value_vars=["Male (Doses Administered)", "Female (Doses Administered)"],
                                      var_name="Gender",
                                      value_name="Doses")

# Plot
plt.figure(figsize=(12,8))
ax = sns.barplot(x="State", y="Doses", hue="Gender", data=gender_data_melted)

plt.xticks(rotation=90)
plt.title("State-wise Male vs Female Vaccination")
plt.xlabel("State")
plt.ylabel("Number of Doses")

for p in ax.patches:
    ax.annotate(format(int(p.get_height()), ','),   # format with commas
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom',
                fontsize=8, rotation=90)

plt.legend(title="Gender")
plt.show()
```
