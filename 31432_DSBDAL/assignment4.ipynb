```python
import seaborn as sns
```

```python
df= sns.load_dataset('titanic')
df.head()
```

## Survival Count

```python
plt.figure()
sns.countplot(x='survived', data=df)
plt.title('Overall Survival Count')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Passenger Count')
plt.show()
```

**Observation:** Most Passengers did not survived

## Survival Based on Passenger Class

```python
plt.figure()
sns.countplot(x='pclass', hue='survived', data=df)
plt.title('Survival Based on Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.show()
```

**Observation**: First Class Passengers survived more then second class and then third class.

## Survival Based on Gender

```python
plt.figure()
sns.countplot(x="sex", hue="survived", data=df)
plt.title("Survival based on Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

**Observation**: Females had a much higher survival rate

## Age Distribution

```python
plt.figure()
sns.histplot(df['age'], bins=30, kde=True)
plt.title("Age based distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

## Survival based on Age distribution

```python
plt.figure()
sns.boxplot(x='survived', y='age',data=df)
plt.title("Age based survival")
plt.xlabel('survived')
plt.ylabel('age')
plt.show()
```

**Observation**: Children and younger passengers survived more

## Fare Daitribution

```python
plt.figure()
sns.histplot(df['fare'], bins=30, kde=True)
plt.title("Fare Distribution")
plt.xlabel('Fare')
plt.ylabel('count')
plt.show()
```

## Survival Based on Fare distribution 

```python
plt.figure()
sns.boxplot( x="survived", y="fare", data=df)
plt.title("Survived Based on Fare")
plt.xlabel("survived")
plt.ylabel("Fare")
plt.show()
```

**Observation**: Higher fare passengers had better survival rate

## Survival Based on Siblings/Spouses

```python
plt.figure()
sns.countplot(x='sibsp', hue='survived', data=df)

plt.title('Survival Based on Siblings/Spouses')
plt.xlabel('Siblings / Spouses')
plt.ylabel('Count')

plt.show()
```

**Observation**: Passengers with small families survived more

## Survival Based on Parents/Children

```python
plt.figure()
sns.countplot(x='parch', hue='survived', data=df)

plt.title('Survival Based on Parents/Children')
plt.xlabel('Parents / Children')
plt.ylabel('Count')

plt.show()
```

Survival based on Embarkation Port

```python
plt.figure()
sns.countplot(x='embarked', hue='survived', data=df)

plt.title('Survival Based on Embarkation Port')
plt.xlabel('Embarked Port')
plt.ylabel('Count')

plt.show()
```

**Observation**: Passengers from Cherbourg (C) had higher survival rate.

```python
plt.figure()
sns.countplot(x="sex", hue="age", data=df)
plt.title("Age and  Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    x="sex",
    y="age",
    hue="survived",
    data=df
)

plt.title("Age Distribution by Gender and Survival Status (Titanic Dataset)")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.legend(title="Survived", labels=["No", "Yes"])

plt.show()
```
