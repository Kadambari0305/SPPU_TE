```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk.tokenize import word_tokenize
```

```python
from nltk.corpus import stopwords
```

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag
```

```python
documents = [
    "Natural Language Processing enables computers to understand human is in  language.",
    "Machine learning improves language models.",
    "Text data processing is important in NLP."
]
```

```python
print(documents)
```

```python
words = word_tokenize(documents[0])
```

```python
import string
stop_word = set(stopwords.words("english"))
filtered_word = [w for w in words if w.lower() not in stop_word and w not in string.punctuation]
```

```python
pos = pos_tag(filtered_word)
```

```python
ps = PorterStemmer()
stemmed = [ps.stem(w) for w in filtered_word]
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w) for w in filtered_word]
```

```python
print("Original Tokens:\n", words)
```

```python
print("Stop word removal: \n", filtered_word)
print("POS Tagging:\n ", pos)
print("Stemming: \n", stemmed)
print("lemmatized words: \n", lemmatized)
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
cv = CountVectorizer()
X = cv.fit_transform(documents)
```

```python
print("Feature Name: \n", cv.get_feature_names_out())
print("TF-Matrix: \n", X.toarray())
```

```python
tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(documents)
```

```python
print("Feature Name:\n", tfidf.get_feature_names_out())
print("TF-IDF Matrix:\n", X_tfidf.toarray())
```

```python
features = tfidf.get_feature_names_out()
idf_values = tfidf.idf_

# Print word importance (IDF)
for word, val in zip(features, idf_values):
    print(word, ":", round(val, 2))
```
