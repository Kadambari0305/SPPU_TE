The Cabin column is removed because it contains many missing values and to reduce
and does not contribute significantly to data analysis.

```python
df.drop(columns=["Cabin"], inplace=True)
```

```python
print(df.info())
```

Convert Name column from object to categorical type to reduce memory usage
and clearly represent it as qualitative data.

```python
df["Name"] = df["Name"].astype("category")
```

```python
#info() represents non null values dataTypes and Memory Usage
print(df.info())
```

Convert the Name column to object (string) data type
converting category to object 
so it can be treated as textual information.

```python
df["Name"] = df["Name"].astype("object")
```

```python
print(df.info())
```

```python
# Selects all numerical columns (integer and float types) from the dataset
df.select_dtypes(include=['int64','float64']).columns  
```

```python
# Selects all categorical/text columns (object type) from the dataset
df.select_dtypes(include=['object']).columns  
```

Mean is used to fill missing values in the Age column because Age is a continuous numerical variable,
and mean imputation preserves the central tendency of the data without removing records
mean is to be used when the data is numerical and continous and has less outliers or less skewed
fillna it replaces the null values(Not a NumberNaN) with other specific values

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
print(df.info())
```

```python
df['Age'].isnull().sum()
```

```python
# Missing values in Embarked are replaced with the most frequent category
#Mode is used to fill missing values in the Embarked column because it is a categorical variable, and mode represents the most frequent category, 
#ensuring data consistency.
#Mode is used mostly when the data or column is category to select most widely used value
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df['Embarked'].isnull().sum()
```

```python
#Shows how many times each category appears after filling.
#value_counts() works for categorical data (display freq of each category)
df['Embarked'].value_counts()
```

```python
# Normalize Age values to the range 0–1 using Min-Max normalization
# df['Age'] - df['Age'].min()) Shifts all age values so the minimum becomes 0.
# (df['Age'].max() - df['Age'].min()) Calculates the range of age values.
df['Age_normalized'] = (df['Age'] - df['Age'].min()) / (df['Age'].max() - df['Age'].min())
```

```python
df[['Age', 'Age_normalized']].head(10)
```

```python
#It converts the categorical Sex column into numerical values binary(0, 1) 
#map() is used for binary(0, 1) conversion
df['Sex_encoded'] = df['Sex'].map({'male': 0, 'female': 1})
```

```python
df['Sex_encoded'].describe()
```

```python
print(df.info())
```

```python
# Convert Embarked categorical column into numerical columns using one-hot encoding
#get_dummies it converts cate to numerical values used when there are multiple category values
#drop_first it reduces the redundancy
df = pd.get_dummies(df, columns=['Embarked'], drop_first=True)
```

```python
print(df.info())
```

```python
df.head()
```

```python
# Save the cleaned and processed dataset into a new CSV file
df.to_csv("processed_data.csv", index=False) 
```
