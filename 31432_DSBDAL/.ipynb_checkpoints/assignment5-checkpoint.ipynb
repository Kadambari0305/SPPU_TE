```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
df= sns.load_dataset('titanic')
df.head()
```

## Age Distribution by Gender and Survival Status

```python


sns.boxplot(
    x="sex",
    y="age",
    hue="survived",
    data=df
)

plt.title("Age Distribution by Gender and Survival Status (Titanic Dataset)")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.legend(title="Survived", labels=["No", "Yes"])

plt.show()
```

**Observation**: Female had higher survival rate thab other, male has less survival rate and children had better survival rate some outliers are present very less edlerly people are survived
