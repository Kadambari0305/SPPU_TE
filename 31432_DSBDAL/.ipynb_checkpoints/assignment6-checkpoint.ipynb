```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "species"
]

iris = pd.read_csv(url, header=None, names=columns)
iris.head()
```

```python
iris.describe()
```

```python
feature_types = pd.DataFrame({
    "Feature": iris.columns,
    "Data Type": iris.dtypes.values
})

feature_types
```

```python
iris.hist(
    figsize=(10, 8),
    bins=20,
    edgecolor="black"
)

plt.suptitle("Histogram of Iris Dataset Features")
plt.show()
```

```python
plt.figure()
sns.boxplot(data = iris.drop('species', axis=1))
plt.title("Boxplot for Iris Dataset Feature")
plt.show()
```

```python
iris.drop("species", axis=1).hist(
    bins=20,
    figsize=(10, 8),
    edgecolor="black"
)

plt.suptitle("Feature Distribution Comparison (Histograms)")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=iris.drop("species", axis=1))
plt.title("Outlier Detection Using Box Plot")
plt.show()
```

```python
numeric_iris = iris.drop("species", axis=1)
```

```python
Q1 = numeric_iris.quantile(0.25)
Q3 = numeric_iris.quantile(0.75)
IQR = Q3 - Q1
```

```python
outliers = (
    (numeric_iris < (Q1 - 1.5 * IQR)) |
    (numeric_iris > (Q3 + 1.5 * IQR))
)
```

```python
for column in numeric_iris.columns:
    print(f"\nOutliers in {column}:")
    print(numeric_iris.loc[outliers[column], column])
```
