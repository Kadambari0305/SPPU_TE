##Summary Statistics Using Adult Census Income Dataset

Objective
The objective of this experiment is to compute summary statistics such as:
- Mean
- Median
- Minimum
- Maximum
- Standard Deviation
for a numeric variable grouped by a categorical variable.

In this analysis:
- **Age Group** is used as the categorical variable.
- **Income** is used as the numeric variable (after conversion).

The dataset used is the Adult Census Income dataset obtained from Kaggle.


```python
import pandas as pd
```

**Dataset Description**
The Adult Census Income dataset contains demographic information about individuals.
For this experiment, the following columns are used:

- age: Age of the individual (numeric)
- income: Income category (`<=50K` or `>50K`) (categorical)

Since statistical operations cannot be directly applied to categorical values,
the income column is converted into numeric form.


The dataset adult.csv is loaded into dataframe

```python
df = pd.read_csv("adult.csv")
df.head()
```

only the columns required for operations are used over here and represented

```python
df = df[['age', 'income']]
df.head()
```

income column is categorical and must be converted to quantitative for analysis
<=50K make it 0
less than 50K make it 1

```python
df['income_numeric'] = df['income'].map({'<=50K' : 0, '>50K': 1})
df.head()
```

Creating Age Groups 
To group individuals meaningfully, age is divided into the following categories:
- 18–25 → Young
- 26–35 → Adult
- 36–50 → Middle-aged
- 51–65 → Senior
- 65+ → Elder
These age groups act as the categorical variable.


```python
bins = [17, 25, 35, 50, 65, 100]
labels = ['young', 'Adult', 'Middle-aged', 'Senior', 'Elder']
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels)
df.head()
```

Computing Summary Statistics
The following statistics are computed for income grouped by age group:
- Mean
- Median
- Minimum
- Maximum
- Standard Deviation

```python
summary_stats = df.groupby('age_group')['income_numeric'].agg(
    Mean='mean',
    Median='median',
    Minimum='min',
    Maximum='max',
    Standard_Deviation='std'
)

summary_stats
```

A list of numeric income values is created for each age group.
This shows how individual income values are distributed within each category.

```python
income_list = df.groupby('age_group')['income_numeric'].apply(list)
income_list
```

## Interpretation of Results
- Younger individuals have a lower proportion of high-income earners.
- Middle-aged and senior groups show a higher likelihood of earning more than 50K.
- Income tends to increase with age up to a certain point.

This demonstrates how categorical grouping helps in understanding data patterns.
## Conclusion
In this experiment:
- The Adult Census Income dataset was analyzed.
- Age groups were treated as a categorical variable.
- Income was converted to numeric form.
- Summary statistics were successfully computed for each age group.
This approach helps in understanding income distribution across different age categories.


```python
import pandas as pd
import numpy as np

df = pd.read_csv("adult.csv")

# Function to assign realistic income
def assign_income(label):
    if label == '<=50K':
        return np.random.randint(15000, 50000)
    else:
        return np.random.randint(50000, 150000)

# Create new income column
df['income_amount'] = df['income'].apply(assign_income)

df[['age', 'income', 'income_amount']].head()
```

```python
bins = [17, 25, 35, 50, 65, 100]
labels = ['18–25', '26–35', '36–50', '51–65', '65+']

df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels)
```

```python
avg_income_age = (
    df.groupby('age_group')['income_amount']
      .mean()
      .reset_index()
)

avg_income_age.rename(
    columns={'income_amount': 'Average Income'},
    inplace=True
)

print(avg_income_age)
```
