```python
import pandas as pd

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag
```

```python

import nltk
```

```python
documents = [
    "This is Assignment is the or not for me so you do",
    "This is Assignment is the or not for me so you do","This is Assignment is the or not for me so you do"]
```

```python
words = word_tokenize(documents[0])
```

```python
stop_word = set(stopwords.words("english"))
filtered_word = [w for w in words if w.lower() not in stop_word ]
```

```python
pos = pos_tag(filtered_word)
```

```python
ps = PorterStemmer()
stemed = [ps.stem(w) for w in filtered_word]

lemmatizer = WordNetLemmatizer()
lemm = [lemmatizer.lemmatize(w) for w in filtered_word]
```

```python
print(words, stemed, lemm)
```

```python
print(stop_word)
```

```python
print(filtered_word)
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
cv = CountVectorizer()
X = cv.fit_transform(documents)
```

```python
print("Feature:", cv.get_feature_names_out())
print("TF:Matrix", X.toarray())
```

```python
tfidf = TfidfVectorizer()
X_array = tfidf.fit_transform(documents)
```

```python
print("Features:\n", tfidf.get_feature_names_out())
print("TF-IDF:\n", X_array.toarray())
```

```python
features = tfidf.get_feature_names_out()
tf_idf = tfidf.idf_
```

```python
for word, val in zip(features, tf_idf):
    print(words, ":", round(val, 2)) 
```
