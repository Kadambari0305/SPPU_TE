Stores the sample document as a string variable to perform text preprocessing operations.

```python
text = """Data science is an interdisciplinary field that uses scientific methods,
processes, algorithms and systems to extract knowledge and insights from data."""
```

```python
import nltk
nltk.download('punkt_tab')
```

text.lower(): Converts all characters to lowercase for uniformity.

word_tokenize(): Breaks the text into individual tokens (words and punctuation).

print(tokens): Displays the list of tokens.

```python
tokens = word_tokenize(text.lower())
print("Tokens:")
print(tokens)
```

```python
import nltk
nltk.download('averaged_perceptron_tagger_eng')
```

pos_tag(): Assigns grammatical tags such as noun (NN), verb (VB), adjective (JJ), etc., to each token.

Helps in understanding the syntactic structure of the text.

```python
from nltk import pos_tag
from nltk.tokenize import word_tokenize

tokens = word_tokenize(text.lower())
print(pos_tag(tokens))
```

loads english stop words and stores them in a set for faster lookup.

This removes common stop words and punctuate symbols and produces meaningful words that contribute to the document

```python
stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word for word in tokens
    if word not in stop_words and word not in string.punctuation
]

print("After Stop Word Removal:")
print(filtered_tokens)
```

PorterStemmer(): Initializes the stemming algorithm.

ps.stem(word): Reduces each word to its stem form.

May produce non-dictionary words.

```python
ps = PorterStemmer()
stemmed_words = [ps.stem(word) for word in filtered_tokens]
print("stemmed Words:")
print(stemmed_words)
```

WordNetLemmatizer(): Initializes the lemmatizer.

lemmatize(): Converts words to their meaningful base form.

More accurate than stemming.

```python
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_tokens]
print("lemmatized Words: ")
print(lemmatized_words)
```

```python
documents=[
    "data science uses data",
    "science extracts knowledge from data",
    "data and science are related"
]
```

Calculates how important a word is across multiple documents.

math.log(): Used to reduce the effect of large document counts.

```python
def inverse_document_frequency(documents):
    idf_dict = {}
    total_docs = len(documents)
    all_words = set(" ".join(documents).split())

    for word in all_words:
        containing_docs = sum(1 for doc in documents if word in doc.split())
        idf_dict[word] = math.log(total_docs / containing_docs)

    return idf_dict
```

```python
idf = inverse_document_frequency(documents)
```

Multiplies TF and IDF values.

Higher TF-IDF score means the word is more important in the document.

```python
def term_frequency(words):
    tf_dict = {}
    total_words = len(words)

    for word in words:
        tf_dict[word] = tf_dict.get(word, 0) + 1

    for word in tf_dict:
        tf_dict[word] = tf_dict[word] / total_words

    return tf_dict


tf = term_frequency(lemmatized_words)
print("TF:")
print(tf)
```

```python
idf = inverse_document_frequency(documents)
print("IDF:")
print(idf)
```

```python
tf_idf = {}

for word in tf:
    tf_idf[word] = tf[word] * idf.get(word, 0)

print("TF-IDF:")
print(tf_idf)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Sample / your documents
documents = [
    "Natural Language Processing is a field of Artificial Intelligence",
    "NLP helps computers understand human language",
    "Text preprocessing is important in NLP"
]

```

```python
#create TF-IDF model
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)
```

```python
#Terms
print("Terms: ")
print(vectorizer.get_feature_names_out())
```

```python
#TF-IDF vaues
print("\nTF-IDF Matrix: ")
print(tfidf_matrix.toarray())
```

```python
# IDF values
idf_values = vectorizer.idf_

for word, idf in zip(vectorizer.get_feature_names_out(), idf_values):
    print(word, ":", idf)
```
